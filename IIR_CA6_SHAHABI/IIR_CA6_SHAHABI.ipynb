````python
import pandas as pd
from collections import Counter
from transformers import BertTokenizer
import matplotlib.pyplot as plt



all_genres = list()
for i in dataset['Genres']:
  for j in range(len(i)):
    j = int(j)
    all_genres.append(i[j])

genre_counts = Counter(all_genres)

top_10_genres = genre_counts.most_common(10)

if top_10_genres:
    genres, counts = zip(*top_10_genres)
    plt.figure(figsize=(10, 6))
    plt.bar(genres, counts, color='skyblue')
    plt.xticks(rotation=45, ha='right')
    plt.xlabel("Genres")
    plt.ylabel("Frequency")
    plt.title("Top 10 Genre Distribution")
    plt.show()
else:
    print("No genres found in the dataset.")

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')


dataset['Summary_Length'] = dataset['Summary'].apply(
    lambda x: len(tokenizer.tokenize(x)) if isinstance(x, str) else 0
)


plt.figure(figsize=(10, 6))
plt.hist(dataset['Summary_Length'], bins=30, alpha=0.7, color='blue')
plt.xlabel("Summary Length (tokens)")
plt.ylabel("Frequency")
plt.title("Summary Length Distribution")
plt.show()

print("Summary Length Statistics:")
print(dataset['Summary_Length'].describe())
````

سپس کتابخانه fd_to_dict_batches رو با توجه به فایل ورکشاپ مربوط به این تمرین تعریف کردم که داده ها به صورت دسته ای تقسیم بشن تا پردازش بهینه تر انجام بشه. در مرحله بعد به سراغ ساخت جدول رفتم و چون میخواستم چند بار کد رو اجرا و دیباگ کنم کدی رو نوشتم که اگه جدولی به نام books وجود داشت اون رو حذف کنه و بعد جدولی به اسم books ساختم که داده های دیتاست و بردارهای امبدینگ درون اون جدول ذخیره بشن.
در مرحله برای بک اپ گرفتن از امبدینگ ها کد را به گوگل درایو متصل کردم و امبدینگ های ساخته شده رو در گوگل درایو ذخیره کردم.


````python

import lancedb
from transformers import AutoTokenizer, AutoModel
import torch
import pandas as pd
import numpy as np
from lancedb.embeddings import get_registry
from lancedb.pydantic import LanceModel, Vector


model_name = "BAAI/bge-small-en-v1.5"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name)

local_db_path = "/content/.lancedb"
db = lancedb.connect(local_db_path)
table_name = "books"
embedding_model = get_registry().get("sentence-transformers").create(name="BAAI/bge-small-en-v1.5", device="cuda")

class StudyNotes(LanceModel):
  ID: str
  Title: str
  Author: str
  Date: str
  Genres: str
  Summary: str = embedding_model.SourceField()
  vector: Vector(embedding_model.ndims()) = embedding_model.VectorField()

def df_to_dict_batches(df: pd.DataFrame, batch_size: int = 128):

    for start_idx in range(0, len(df), batch_size):
        end_idx = start_idx + batch_size
        batch_dicts = df.iloc[start_idx:end_idx].to_dict(orient="records")
        yield batch_dicts

if table_name in db.table_names():
    db.drop_table(table_name)


tbl = db.create_table(
    table_name,
    data=df_to_dict_batches(dataset, batch_size=10),
    schema=StudyNotes
)
print(f"Table '{table_name}' created successfully!")
from google.colab import drive
import os
import shutil

drive.mount('/content/drive')

local_lancedb_path = ".lancedb"


drive_lancedb_path = "/content/drive/My Drive/lancedb_backup"


if os.path.exists(drive_lancedb_path):
    print("Restoring LanceDB from Google Drive...")
    if os.path.exists(local_lancedb_path):
        shutil.rmtree(local_lancedb_path)
    shutil.copytree(drive_lancedb_path, local_lancedb_path)
else:
    print("No LanceDB backup found in Google Drive.")

if not os.path.exists(drive_lancedb_path):
    print("Saving LanceDB to Google Drive...")
    shutil.copytree(local_lancedb_path, drive_lancedb_path)
````

در این قسمت سعی کردم کد بنویسم که میزان اشتراک ژانر بین هر کتاب موجود در دیتا ست را اندازه گیری کند و در یک دیتافریم جدید ذخیره کند. بعد از ایمپورت کردن کتابخانه های مورد نیاز تابع calculate_overlap رو برای محاسبه اشتراک بین ژانر ها تعریف کردم. این تابع دو لیست ژانر ۱ و ۲ را به عنوان ورودی دریافت می کند و مقدار اشتراک ژانر را به این صورت محاسبه میکند: ابتدا تعداد ژانر های مشترک را حساب کرده و سپس نسبت تعداد ژانرهای مشترک را به کل ژانر های هر لیست محاسبه میکند و مقدار حداقل آن را بر میگرداند و خروجی مقدار اشتراک بین ۰ تا ۱ است. سپس با ایتفاده از این تابع اشتراک را برای همه ترکیب های دیتا ست محاسبه کردم و نتایج را در یک دیتافریم ذخیره کردم و در نهایت دیتا فریم رو چاپ کردم.

````python
import pandas as pd
from itertools import combinations


def calculate_overlap(genres1, genres2):
    common_genres = set(genres1).intersection(set(genres2))
    overlap = min(len(common_genres) / len(genres1), len(common_genres) / len(genres2))
    return overlap

results = []
for (id1, genres1), (id2, genres2) in combinations(zip(dataset['ID'], dataset['Genres']), 2):
    overlap = calculate_overlap(genres1, genres2)
    results.append({'id_1': id1, 'id_2': id2, 'overlap': overlap})

overlap_df = pd.DataFrame(results)

print(overlap_df.head())
````

````python

def categorize_relation(row):
    if row['overlap'] > 0.5 and row['relevance'] > 0.8:
        return 'relevant'
    elif row['overlap'] < 0.2 and row['relevance'] < 0.7:
        return 'nonrelevant'
    else:
        return 'unkwon'

combined_df['category'] = combined_df.apply(categorize_relation, axis=1)
print(combined_df)
````

````python
import pandas as pd
related_books = combined_df[combined_df['category'] == 'relevant']
unrelated_books = combined_df[combined_df['category'] == 'nonrelevant']

related_counts = related_books.groupby('id_1').size().reset_index(name='related_count')
unrelated_counts = unrelated_books.groupby('id_1').size().reset_index(name='unrelated_count')

book_counts = pd.merge(related_counts, unrelated_counts, on='id_1', how='inner')

valid_books = book_counts[
    (book_counts['related_count'].between(10, 20)) &
    (book_counts['unrelated_count'].between(10, 20))
]

selected_books = valid_books.sort_values(by='related_count', ascending=False).head(5)['id_1'].tolist()

print("Selected book IDs (most relevant):", selected_books)

filtered_related_books = related_books[related_books['id_1'].isin(selected_books)]
filtered_unrelated_books = unrelated_books[unrelated_books['id_1'].isin(selected_books)]

print("Relevant books for selected IDs:")
print(filtered_related_books)

print("Non-relevant books for selected IDs:")
print(filtered_unrelated_books)


relevant_output = filtered_related_books[['id_1', 'id_2', 'relevance', 'overlap']]
nonrelevant_output = filtered_unrelated_books[['id_1', 'id_2', 'relevance', 'overlap']]

print("\nRelevant books output:")
print(relevant_output)

print("\nNon-relevant books output:")
print(nonrelevant_output)
````

````python
import os
import pandas as pd
from typing import List
from pydantic import BaseModel, Field
from langchain.chat_models import ChatOpenAI
from langchain.prompts import ChatPromptTemplate
from langchain.schema import HumanMessage
from langchain.output_parsers import PydanticOutputParser


PROVIDER_BASE_URL = "https://api.together.xyz/v1"
MODEL_NAME = "meta-llama/Meta-Llama-3.1-70B-Instruct-Turbo"
os.environ["TOGETHER_API_KEY"] = os.getenv("TOGETHER_API_KEY", "")
api_token = os.getenv("TOGETHER_API_KEY")

model = ChatOpenAI(
    base_url=PROVIDER_BASE_URL,
    api_key=api_token,
    model=MODEL_NAME,
    temperature=0.4,
)

class GeneratedQuestion(BaseModel):
    type: str = Field(description="Type of question (general or detailed)")
    question: str = Field(description="Generated question")

class GeneratedQuestionsOutput(BaseModel):
    questions: List[GeneratedQuestion] = Field(description="List of generated questions.")


output_parser = PydanticOutputParser(pydantic_object=GeneratedQuestionsOutput)


prompt_template = ChatPromptTemplate.from_template("""
You are an intelligent assistant. Your task is to generate meaningful questions based on book summaries.
Here is the main book summary:
{main_summary}

Here are the summaries of related books:
{related_summaries}

Generate two types of questions:
1. Two general questions about common themes or shared topics between the main book and related books.
2. Three detailed questions about specific details of the main story or related stories.

{format_instructions}
""")


format_instructions = output_parser.get_format_instructions()


def generate_queries(main_summary: str, related_summaries: List[str], retries: int = 3) -> List[GeneratedQuestion]:

    prompt = prompt_template.format(
        main_summary=main_summary,
        related_summaries="\n".join(related_summaries),
        format_instructions=format_instructions
    )
    for attempt in range(retries):
        try:
            response = model([HumanMessage(content=prompt)])
            print(f"Raw model output:\n{response.content}")

            parsed_output = output_parser.parse(response.content)
            return parsed_output.questions
        except Exception as e:
            print(f"Error in attempt {attempt + 1}: {e}")
    raise Exception(f"Failed to generate queries after {retries} attempts.")


def process_books(df: pd.DataFrame, output_file: str = None):
    all_queries = []
    grouped = df.groupby("id_1")
    for book_id, group in grouped:
        main_summary = group.iloc[0]['Summary_1']
        related_summaries = group['Summary_2'].head(5).tolist()
        print(f"Processing book ID: {book_id}")
        try:

            queries = generate_queries(main_summary, related_summaries)

            for query in queries:
                all_queries.append({
                    "Book_ID": book_id,
                    "Type": query.type,
                    "Question": query.question
                })
            print(f"Generated queries for book ID {book_id}: {len(queries)} questions")
        except Exception as e:
            print(f"Failed to generate queries for book ID {book_id}: {e}")


    queries_df = pd.DataFrame(all_queries)


    if queries_df.empty:
        print("No queries were generated. The DataFrame is empty.")
    else:
        print(f"Generated queries DataFrame:\n{queries_df}")

    if output_file:
        queries_df.to_csv(output_file, index=False)
        print(f"Queries saved to {output_file}")

if __name__ == "__main__":

    df = final_merged
    process_books(df, output_file="/content/drive/My Drive/IR_CA6/generated_queries.csv")
````

````python
# WRITE YOUR CODE HERE


import pandas as pd
file_path = "/content/drive/My Drive/IR_CA6/generated_answers.csv"
file = pd.read_csv(file_path)
file_df = pd.DataFrame(file)


import pandas as pd




scores_df = pd.DataFrame(columns=['Question', 'Relevant documents', 'Answer',
                                  'Factual Accuracy', 'Relevance', 'Coherence', 'Factual Consistency'])

while True:
    print("\nCurrent Questions:")
    print(file_df[['Question']])
    row_number = input("\nEnter the row number of the question to review (or type 'exit' to quit): ")


    if row_number.lower() == 'exit':
        break

    try:
        row_number = int(row_number)
        if row_number < 0 or row_number >= len(file_df):
            print("Invalid row number. Please try again.")
            continue


        selected_question = file_df.loc[row_number, 'Question']
        relevant_docs = file_df.loc[row_number, 'Relevant documents']
        answer = file_df.loc[row_number, 'Answer']

        print(f"\nSelected Question:\n{selected_question}")
        print(f"\nRelevant Documents:\n{relevant_docs}")
        print(f"\nAnswer:\n{answer}")


        factual_accuracy = float(input("Enter score for Factual Accuracy (1-5): "))
        relevance = float(input("Enter score for Relevance (1-5): "))
        coherence = float(input("Enter score for Coherence (1-5): "))
        factual_consistency = float(input("Enter score for Factual Consistency (1-5): "))


        scores_df = pd.concat([
            scores_df,
            pd.DataFrame([{
                'Question': selected_question,
                'Relevant documents': relevant_docs,
                'Answer': answer,
                'Factual Accuracy': factual_accuracy,
                'Relevance': relevance,
                'Coherence': coherence,
                'Factual Consistency': factual_consistency
            }])
        ], ignore_index=True)

        print("\nScores saved successfully!")

    except ValueError:
        print("Invalid input. Please enter a valid number.")
        continue


print("\nFinal Scores DataFrame:")
print(scores_df)

````

````python
def display_all_scores_info(scores_df):
    for index, row in scores_df.iterrows():
        selected_question = row['Question']
        relevant_docs = row['Relevant documents']
        answer = row['Answer']
        factual_accuracy = row['Factual Accuracy']
        relevance = row['Relevance']
        coherence = row['Coherence']
        factual_consistency = row['Factual Consistency']
        avg_score = (factual_accuracy + relevance + coherence + factual_consistency) / 4

        print(f"\n=== Question {index + 1} ===")
        print(selected_question)
        print("\n=== Relevant Documents ===")
        print(relevant_docs)
        print("\n=== Generated Answer ===")
        print(answer)
        print("\n=== Scores ===")
        print(f"Factual Accuracy: {factual_accuracy}")
        print(f"Relevance: {relevance}")
        print(f"Coherence: {coherence}")
        print(f"Factual Consistency: {factual_consistency}")
        print(f"Average Score: {avg_score:.2f}")
        print("-" * 50)

display_all_scores_info(scores_df)
````

<h4 dir="rtl" style="font-family: Vazir; width: 85%;">صداقت علمی:</h4> <ul dir="rtl" style="font-family: Vazir; width: 85%; font-size: 16px;"> <li>ما نوت‌بوک‌های تعداد مشخصی از دانشجویان که به صورت تصادفی انتخاب می‌شوند، بررسی خواهیم کرد. این بررسی‌ها اطمینان حاصل می‌کنند که کدی که نوشتید واقعاً پاسخ‌های موجود در نوت‌بوک شما را تولید می‌کند. اگر پاسخ‌های صحیح را در نوت‌بوک خود بدون کدی که واقعاً آن پاسخ‌ها را تولید کند تحویل دهید، این یک مورد جدی از عدم صداقت علمی محسوب می‌شود.</li> <li>ما همچنین بررسی‌های خودکاری را برای تشخیص سرقت علمی در نوت‌بوک‌های کولب انجام خواهیم داد. کپی کردن کد از دیگران نیز یک مورد جدی از عدم صداقت علمی محسوب می‌شود.</li> </ul>

<h4 dir="rtl" style="font-family: Vazir; width: 85%;">توضیحات تکمیلی:</h4> <ul dir="rtl" style="font-family: Vazir; width: 85%; font-size: 16px;">
<li>
خوانایی و دقت بررسی‌ها در گزارش نهایی از اهمیت ویژه‌ای برخوردار است. به تمرین‌هایی که به صورت کاغذی تحویل داده شوند یا به صورت عکس در سایت بارگذاری شوند، ترتیب اثری داده نخواهد شد.</li>
<li>
 همه‌ی کدهای پیوست گزارش بایستی قابلیت اجرای مجدد داشته باشند. در صورتی که برای اجرا مجدد آن‌ها نیاز به تنظیمات خاصی می‌باشد، بایستی تنظیمات مورد نیاز را نیز در گزارش خود ذکر کنید.  دقت کنید که  تمامی کدها باید توسط شما اجرا شده باشند و نتایج اجرا در فایل کدهای ارسالی مشخص باشد. به کدهایی که نتایج اجرای آن‌ها در فایل ارسالی مشخص نباشد نمره‌ای تعلق نمی‌گیرد.
</li>
<li>
