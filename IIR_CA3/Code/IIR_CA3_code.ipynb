الف

<div style="direction: rtl; text-align: right;">
در این کد ابتدا کتابخانه های مورد نیاز ایمپورت شده اند. کتابخانه pandas جهت پردازش داده های جدولی، کتابخانه nlkt برای توکن سازی، حذف استاپ ورد ها و یافتن stem کلمات. از کتابخانه beautifulsoup هم برای حذف تگ  های html استفاده کردم.
ابتدا لیست استاپ ورد های انگلیسی و مدل های مورد نیاز برای توکنایز کردن متن دانلود شده است.
سپس دیتاست نظرات کاربران را از مسیر مشخص شده آپلود کرده ام. در مرحله بعد لیست استاپ وردها را به صورت set ذخیره کردم. سپس از porterstemmer برای پیدا کردن stem کلمات را جدا کردم.
در مرحله بعد با استفاده از preprocess_text ابتدا تگ های html را در متن حذف کردم سپس متن ورودی را به کوچکترین شکل ممکن توکنایز کردم و کلمات غیر عددی بدون استاپ ورد را استخراج کردم.
در مرحله بعد نظرات مثبت و منفی با استفاده از فیلتر کردن بر اساس ستون sentiment استخراج می‌شوند و در دو DataFrame جداگانه ذخیره می‌شوند (positive_reviews و negative_reviews).
در پایان نظرات مثبت چاپ میشوند.
نظرات منفی هم در رشته کد بعدی چاپ میشوند.
تعداد نظرات مثبت و منفی برابر ۱۲۵۰۰ است.

```python
import pandas as pd
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
from bs4 import BeautifulSoup  

nltk.download('stopwords')
nltk.download('punkt')


data_path = '/Users/erfanshahabi/.cache/kagglehub/datasets/lakshmi25npathi/imdb-dataset-of-50k-movie-reviews/versions/1/IMDB Dataset.csv'


df = pd.read_csv(data_path)


stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()


def clean_html(text):
    return BeautifulSoup(text, "html.parser").get_text()


def preprocess_text(text):
   
    text = clean_html(text)
   
    tokens = word_tokenize(text.lower())
    filtered_tokens = [stemmer.stem(word) for word in tokens if word.isalnum() and word not in stop_words]
    return filtered_tokens


df['processed_text'] = df['review'].apply(preprocess_text)


positive_reviews = df[df['sentiment'] == 'positive']
negative_reviews = df[df['sentiment'] == 'negative']


print(positive_reviews)
```

<div style="direction: rtl; text-align: right;">
در این کد سعی شده ۵۰ کلمه پرکاربرد در نظرات استخراج شوند. 
ابتدا counter از کتابخانع collections ایمپورت شده است که از آن برای شمارش فراوانی کلمات استفاده می شود.
سپس تابع get_top_words تعریف شده است که نظرات و تعداد کلمات پرکاربرد درخواستی را به عنوان ورودی می گیرد.
ابتدا تمامی کلمات موجود در نظرات جمع‌آوری می‌شوند. این کار با استفاده از یک لیست فشرده انجام می‌شود که برای هر نظر، تمام کلمات آن را استخراج می‌کند.
سپس از Counter برای شمارش فراوانی هر کلمه استفاده شده و n کلمه پرکاربردتر را برمی‌گرداند.
در مرحله بعد با فراخوانی تابع get_top_words برای نظرات مثبت (positive_reviews['processed_text']) و نظرات منفی (negative_reviews['processed_text'])، ۵۰ کلمه پرکاربرد از هر دسته استخراج می‌شود.
برای ذخیره کلمات پرکاربرد در فایل‌ها، دو فایل متنی به نام‌های top_words_positive.txt و top_words_negative.txt ایجاد می‌شود و کلمات در آن ها ذخیره میشوند که در فایل ارسالی در elearn قابل مشاهده هستند.

```python
from collections import Counter

def get_top_words(reviews, n=50):
    all_words = [word for review in reviews for word in review]
    return Counter(all_words).most_common(n)


top_positive_words = get_top_words(positive_reviews['processed_text'])
top_negative_words = get_top_words(negative_reviews['processed_text'])


with open('top_words_positive.txt', 'w') as pos_file, open('top_words_negative.txt', 'w') as neg_file:
    print("Top 50 Words in Positive Reviews:")
    for word, count in top_positive_words:
        print(f"{word}: {count}")
        pos_file.write(f"{word}: {count}\n")
    
    print("\nTop 50 Words in Negative Reviews:")
    for word, count in top_negative_words:
        print(f"{word}: {count}")
        neg_file.write(f"{word}: {count}\n")
```

<div style="direction: rtl; text-align: right;">
در این کد تلاش کردم ۵۰ جفت کلمه پر تکرار در نظرات منفی و مثبت را استخراج کنم.
ابتدا تابع get_top_bigrams رو تعریف کردم که این تابع به ازای هر نظر در دیتاست تمام بای گرم های موجود در نظر را استخراج میکند. برای این کار از list comprehension استفاده شده است که تمامی بای گرم‌ ها را برای هر نظر استخراج می‌کند سپس از کانتر برای شمارش فراوانی جفت کلمات استفاده می کند. در نهایت با استفاده از most_common(50)، ۵۰ جفت کلمه پر کاربرد را استخراج می کند.
سپس با استفاده از این تابع ۵۰ جفت کلمه پر کاربرد در نظرات مثبت و منفی را محاسبه کرده و در فایل مربوطه ذخیره میکند و ان ها را چاپ هم میکند.

```python
def get_top_bigrams(reviews, n=50):
    all_bigrams = [tuple(review[i:i+2]) for review in reviews for i in range(len(review)-1)]
    return Counter(all_bigrams).most_common(n)


top_positive_bigrams = get_top_bigrams(positive_reviews['processed_text'])
top_negative_bigrams = get_top_bigrams(negative_reviews['processed_text'])

with open('top_bigrams_positive.txt', 'w') as pos_file, open('top_bigrams_negative.txt', 'w') as neg_file:
    print("\nTop 50 Bigrams in Positive Reviews:")
    for bigram, count in top_positive_bigrams:
        print(f"{bigram}: {count}")
        pos_file.write(f"{bigram}: {count}\n")
    
    print("\nTop 50 Bigrams in Negative Reviews:")
    for bigram, count in top_negative_bigrams:
        print(f"{bigram}: {count}")
        neg_file.write(f"{bigram}: {count}\n")
```

در این کد سعی شده گراف رابطه بین محصولات و کاربران رسم شود که کاربران و محصولات نود های آن هستند و یال های آن با استفاده از نظراتی که کاربر برای آن محصول ثبت کرده است رسم میشود.
ابتدا کتابخانه های پانداس و networkx ایمپورت شده اندو کتابخانه networkx برای ساخت و مدیریت استفاده می شود.
سپس فایل دیتا ست آدرس دهی شده و فایل خوانده می شود و ستون های مربوط به کاربران، محصپلات و امتیاز کاربر به محصول جدا شدند. در مرحله بعد یک گراف جهت دار ایجاد میشود که در این گراف کاربران و محصولات به عنوان نود های جداگانه خواهند بود.
سپس براای هر ردیف داده ها که نمایانگر یک نظر هستند یک یال بین کاربر و محصول کشیده می شود که وزن آن یال score است.
گزاف رسم شده باید گرافی جهت دار باشد چرا که یال ها از کاربر به محصول کشیده می شود همچنین این گراف وزن دار است و وزن یال ها برابر score میباشند و این گراف شامل دو نوع نود کاربر و محصول است.

```python
import pandas as pd
import networkx as nx


data_path = '/Users/erfanshahabi/.cache/kagglehub/datasets/snap/amazon-fine-food-reviews/versions/2/Reviews.csv'
df = pd.read_csv(data_path)

df = df[['UserId', 'ProductId', 'Score']]


G = nx.DiGraph()


for _, row in df.iterrows():
    user = row['UserId']
    product = row['ProductId']
    score = row['Score']
    G.add_edge(user, product, weight=score)


print(f"Graph created with {G.number_of_nodes()} nodes and {G.number_of_edges()} edges.")
```

ب و ج

<div style="direction: rtl; text-align: right;">
در این کد با استفاده از کتابخانه networkx الگوریتم hits ۱۰ بار تکرار شده و مقادیر hub و authorities برای هر نود محاسبه میشود.
سپس در رشته کد بعدی تعداد یال هایی که به هر نود وارد و یا از هر نود خارج میش.وند را محاسبه کردم. این مقادیر به صورت دیکشنری تبدیل شده‌اند که در آن هر گره به عنوان کلید و درجه ورودی یا خروجی به عنوان مقدار ذخیره می‌شود. سپس با استفاده  از مقادیر هاب و اَثرَیتی که قبلاً محاسبه شده‌اند (از طریق تابع nx.hits) استفاده شده است. برای هر گره در شبکه G، نمرات هاب و authority به ترتیب در hubs_vector و authorities_vector ذخیره می‌شوند. و بردار های ورودی و خروجی به هر نود نیز ایجاد میشوند. با استفاده از متد corrcoef در کتابخانه نامپای میزان همبستگی بین بردار هاب و بردار یال های خروجی و بین بردار authority و بردار یال های ورودی محاسبه شده که مقادر 0.19, 0.29 شده و چون این مقادیر کم هستند نشان دهنده این است که بردار های هاب و authority باید تغییر کنند.
این تغییر می تواند با تنظیم وزن و نرمال سازی بردارها، افزایش تکرار و یا تغییر ساختار گراف باشد.
این قسمت از کد با استفاده از راهنمایی های chatgpt نوشته شده است که پرامپت آن در ادامه آمده است.
پرامپت: https://chatgpt.com/share/673a2523-2518-800d-99a8-a59d7c6acc84
<br><br>
برای قسمت ج من فکر میکنم اگر Hub یک محصول بالا باشد (یعنی بسیاری از کاربران به آن ارجاع داده‌اند یا آن را بازدید کرده‌اند)، احتمالاً این محصول برای کاربران جذاب است و authority بیشتری نیز خواهد داشت (یعنی این محصول توسط کاربران بیشتری خریداری یا بررسی شده است).
authority یک کاربر بالا (یعنی کاربری که بیشترین تعداد محصول را ارزیابی یا خریداری کرده) نشان‌دهنده این است که این کاربر احتمالاً یک hub خوب نیز است، زیرا به احتمال زیاد محصولات بیشتری را بررسی یا خریداری کرده و در نتیجه به دیگر کاربران ارجاع می‌دهد به همین منظور من فکر میکنم اگر همبستگی زیاد باشد، نشان‌دهنده این است که محصولاتی که hub بالایی دارند، بیشتر توسط کاربران تکرار می‌شوند و  کاربران با authority بالا، بیشتر به محصولات مراجعه یا آنها را تکرار می‌کنند. از آنجایی که با محاسبه همبستگی فهمیدیم این مقدار پایین است پس اگر بتوانیم این همبستگی را افزایش دهیم، تاثیر میزان تکرار محصولات و کاربران بر دو بردار hubs , authorities بیشتر خواهد شد.

```python

hubs, authorities = nx.hits(G, max_iter=10)


print("\nTop 10 Hubs:")
top_hubs = sorted(hubs.items(), key=lambda x: x[1], reverse=True)[:10]
for node, score in top_hubs:
    print(f"{node}: {score}")


print("\nTop 10 Authorities:")
top_authorities = sorted(authorities.items(), key=lambda x: x[1], reverse=True)[:10]
for node, score in top_authorities:
    print(f"{node}: {score}")
```

```python

in_degrees = dict(G.in_degree())  
out_degrees = dict(G.out_degree()) 
import numpy as np

hubs_vector = np.array([hubs[node] for node in G.nodes()])
authorities_vector = np.array([authorities[node] for node in G.nodes()])


in_degree_vector = np.array([in_degrees[node] if node in in_degrees else 0 for node in G.nodes()])
out_degree_vector = np.array([out_degrees[node] if node in out_degrees else 0 for node in G.nodes()])


correlation_hub_outdegree = np.corrcoef(hubs_vector, out_degree_vector)[0, 1]
correlation_authority_indegree = np.corrcoef(authorities_vector, in_degree_vector)[0, 1]

print(f"Correlation between Hub Scores and Out-degree: {correlation_hub_outdegree:.2f}")
print(f"Correlation between Authority Scores and In-degree: {correlation_authority_indegree:.2f}")
print(hubs_vector)
print(authorities_vector)
```

د

<div style="direction: rtl; text-align: right;">
در این کد من از متد pagerank کتابخانه networkx استفاده کردم.
ابتدا با استفاده از این متد پیج رنک را بر حسب آلفا برابر ۰.۸۵ با تکرار ۱۰ بار محاسبه کردم.
آلفا میتواند از ۰ تا ۱ باشد ولی معمولا مقدار آن را ۰.۸۵ در نظر میگیرند.
در نهایت، تابع nx.pagerank() نمرات PageRank برای تمامی گره‌ها در گراف G را محاسبه کرده و در دیکشنری pagerank_scores ذخیره می‌کند. هر نود در گراف به یک نمره PageRank مربوط می‌شود که نشان‌دهنده اهمیت آن نود است. در پایان ابتدا، لیستی از گره‌ها و نمرات PageRank آن‌ها که در df['ProductId'] (که حاوی شناسه‌های محصولات است) موجود هستند، استخراج می‌شود.
سپس این لیست بر اساس نمرات PageRank به ترتیب نزولی مرتب می‌شود.
با استفاده از [:10]، ۱۰  محصول با بیشترین نمره PageRank انتخاب می‌شود و این ۱۰ محصول چاپ میشوند.

```python

pagerank_scores = nx.pagerank(G, alpha=0.85, max_iter=10)


print("\nTop 10 Products by PageRank Scores:")
top_products = sorted(
    [(node, score) for node, score in pagerank_scores.items() if node in df['ProductId'].values],
    key=lambda x: x[1],
    reverse=True
)[:10]

for product, score in top_products:
    print(f"Product ID: {product}, PageRank Score: {score}")

with open('top_pagerank_products.txt', 'w') as pr_file:
    for product, score in top_products:
        pr_file.write(f"Product ID: {product}, PageRank Score: {score}\n")
```

سپس یک دیکشنری‌ به نام personalization ایجاد کردم که برای تمامی نودهای گراف (G) مقدار اولیه ۰ را به‌عنوان نمره شخصی‌سازی قرار می‌دهد.
در مرحله بعد برای هر محصولی که توسط کاربر انتخاب‌شده ارزیابی شده، مقدار آن محصول در دیکشنری personalization به ۱ تغییر داده می‌شود. و سپس  الگوریتم Topic Sensitive PageRank با استفاده از پارامتر personalization (که به محصولات ارزیابی‌شده توسط کاربر اختصاص داده شده است) اجرا می‌شود. این باعث می‌شود که در این الگوریتم، محصولات مرتبط با کاربر (بر اساس ارزیابی‌ها) اولویت بالاتری داشته باشند.
پس از محاسبه نمرات page rank، ده محصول برتر انتخاب می‌شوند. این محصولات باید در دیتافریم df موجود باشند تا در فهرست نهایی قرار گیرند. در مرحله آخر این ۱۰ محصول چاپ شده و سپس نمراتی که کاربر به این ۱۰ محصول داده نیز چاپ میشود. نمراتی که برای هر محصول نمایش داده شده نشان میدهد این الگوریتم تا حد زیادی محصولات با نمرات بالا را در رتبه های بالاتری چاپ کرده، بجز ۲ محصول که امتیاز ۲ دارند، امتیاز بقیه محصولات بالای ۳ بوده و غالب این محصولات هم امتیاز برابر ۵ دارند.


```python

import random

user_review_counts = df['UserId'].value_counts()


users_with_many_reviews = user_review_counts[user_review_counts > 10].index


selected_user = random.choice(users_with_many_reviews)
print(f"Randomly Selected User ID: {selected_user}")

personalization = {node: 0 for node in G.nodes()}
for _, row in df[df['UserId'] == selected_user].iterrows():
    product = row['ProductId']
    if product in G.nodes():
        personalization[product] = 1


topic_sensitive_pagerank = nx.pagerank(G, alpha=0.85, personalization=personalization)


top_products_user = sorted(
    [(node, score) for node, score in topic_sensitive_pagerank.items() if node in df['ProductId'].values],
    key=lambda x: x[1],
    reverse=True
)[:10]

print("\nTop 10 Products for Selected User by Topic-sensitive PageRank:")
for product, score in top_products_user:
    print(f"Product ID: {product}, PageRank Score: {score}")

print("\nUser's Ratings for Top 10 Products:")
user_ratings = df[(df['UserId'] == selected_user) & (df['ProductId'].isin([p[0] for p in top_products_user]))]

for _, row in user_ratings.iterrows():
    print(f"Product ID: {row['ProductId']}, User Rating: {row['Score']}")
```
